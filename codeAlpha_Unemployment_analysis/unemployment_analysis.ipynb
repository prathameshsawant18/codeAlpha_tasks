```python
import zipfile
import pandas as pd

with zipfile.ZipFile('/content/archive (1).zip', 'r') as zip_ref:
    zip_ref.extractall('/content/')

df = pd.read_csv('/content/Unemployment in India.csv')
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df.columns = df.columns.str.strip()
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Date', y='Estimated Unemployment Rate (%)', hue='Area')
plt.title('Unemployment Rate Trend in India (Rural vs Urban)')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.show()

print(df.describe())
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Estimated Unemployment Rate (%)', y='Region', ci=None)
plt.title('Average Unemployment Rate by State')
plt.xlabel('Mean Unemployment Rate (%)')
plt.show()
```
